這是一份專為 Windows (VS Code) 開發環境量身打造的日常操作指南。

## 一、 虛擬環境日常操作 🐍
每次重新打開專案時，請依據以下步驟檢查並啟用環境。

### 1. 啟動虛擬環境
在 VS Code 終端機 (PowerShell) 中輸入以下指令。啟動成功後，終端機最前方會出現 `(.venv)` 字樣：

````python
# Windows (PowerShell) 啟動指令
.\.venv\Scripts\Activate.ps1
````

### 2. 安裝與管理第三方套件
必須在看得到 `(.venv)` 的狀態下執行，套件才會獨立安裝在此專案中：

````python
# 安裝特定套件 (以 requests 套件為例)
pip install requests

# 檢查目前環境中已安裝的所有套件清單
pip list
````

### 3. 關閉虛擬環境
若想離開目前的虛擬環境，直接在終端機輸入：

````python
deactivate
````

## 二、 Git / GitHub 日常更新黃金三步驟 🚀
當您在 `app.py` 寫完新程式碼、存檔後，想要同步進度到 GitHub，請**依序單行**在終端機執行以下指令：

````python
# 步驟 1：把所有修改過的檔案收集到暫存區（注意後面有一個點）
git add .

# 步驟 2：提交本次進度並寫下修改紀錄（雙引號內可自訂說明）
git commit -m "更新了 app.py 的功能"

# 步驟 3：直接推送到 GitHub 雲端倉庫（之後更新只需輸入簡短的 git push）
git push
````

## 三、 進階配置：`.gitignore` 檔案 🛑
為了避免把幾百 MB 的虛擬環境設定檔推送到 GitHub 導致專案臃腫，我們需要建立一個忽略清單。

**操作方式：**
1. 在專案根目錄下新建一個名為 `.gitignore` 的檔案（前面有一個點）。
2. 在該檔案中寫入以下內容並存檔：
```text
.venv/
project/
__pycache__/
```
